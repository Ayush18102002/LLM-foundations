## Recap


```python
from transformers import AutoTokenizer
checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
tokenzer = AutoTokenizer.from_pretrained(checkpoint)
sequence = "I've been waiting for a HuggingFace course my whole life."
tokens = tokenzer.tokenize(sequence)
print(tokens)
```

```python
ids= tokenzer.convert_tokens_to_ids(tokens)
print(ids)
```

```python
# will pad the sequence up to the maximum sequence length
model_inputs = tokenzer(sequence,padding="longest")
print(model_inputs)
```

```python
# will pad the sequence up to the model max length
model_inputs = tokenzer(sequence, padding="max_length")
print(model_inputs)
```

```python
# will pad the sequence up to specified max length
model_inputs = tokenzer(sequence, padding="max_length",mx_length=8)
print(model_inputs)
```

```python
# truncate
sequence = ["I've like is very very very very very very very very very much toooo"]

trail = tokenzer.tokenize(sequence)
print(trail)

# Will truncate the sequences that are longer than the model max length
model_inputs = tokenzer(sequence,truncation=True)
print(model_inputs)
```

```python
# will truncate the sequence that are longer than the specified max length
model_inputs = tokenzer(sequence,max_length=8,truncation=True)
print(model_inputs)
```

```python
sequence = ["I've been waiting for my friend at metro station"]

# return pytorch tensors

model_inputs = tokenzer(sequence, padding=True, return_tensors="pt")
print(model_inputs)

# return numpy tensor

model_inputs = tokenzer(sequence, padding=True, return_tensors="np")
print(model_inputs)
```

```python
sequence = "nice too meet you!"
model_inputs = tokenzer(sequence)
print(model_inputs["input_ids"])

tokens = tokenzer.tokenize(sequence)
ids = tokenzer.convert_tokens_to_ids(tokens)
print(ids)
```

```python
print(tokenzer.decode(model_inputs["input_ids"]))
print(tokenzer.decode(ids))
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification

checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
model = AutoModelForSequenceClassification.from_pretrained(checkpoint)
sequence = ["I've been waiting for a LLM course my whole life.", "So have I!"]

tokens = tokenzer(sequence, padding=True, truncation=True, return_tensors="pt")
output= model(**tokens)
print(output)
```
